```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

%matplotlib inline
plt.style.use('default')
```

## Matplotlib Architecture

### Key Concepts:
- **Figure**: **The top level container (entire window/page)**
- **Axes**: **The actual plot area (data is drawn)**
- **Axis**: **The number line objects (x-axis, y-axis)**
- **Think of Figure as a Canvas and Axes as indiviual paintings on that canvas**


```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.plot(y, x)

plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Simple Line Plot')
plt.show()
```

```python
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_title('Line Plot (Object Oriented)')
plt.show()
```

## Basic Line Plots

```python
x = np.linspace(0, 10, 100) # 100 points from 0 to 10
y = np.sin(x)
# Simple line plot
plt.figure(figsize =(8, 5)) # Figsize =(width, height) in inches
plt.plot(x, y)
plt.xlabel('X Values')
plt.ylabel('Sin(x)')
plt.title('Sin Wave')
plt.grid(True) # Add grid for easier reading
plt.show()
```

```python
# Multiple lines on the same plot
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)
plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='Sin(x)')
plt.plot(x, y2, label='Cos(x)')
plt.plot(x, y3, label='Sin(x) * Cos(x)')
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.title('Multiple Functions')
plt.legend() # Show legend
plt.grid(True, alpha=0.3) # Alpha controls transparency
plt.show()

```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]

plt.plot(x, y1, label='Squared')
plt.plot(x, y2, label='Linear')

plt.legend()

plt.show()
```

## Customizing Plots: Colors, Styles and Markers
### Matplotlib Offers extensive customization options to make your plots visually appealing and informative.

```python
x = np.linspace(0, 10, 20)
# Different line styles and markers
plt.figure(figsize=(12,6))

plt.plot(x, x, 'b-', label='Solid Line', linewidth=2)
plt.plot(x, x+1, 'r--', label='Dashed Line', linewidth=2)
plt.plot(x, x+2, 'g-', label='Dash-Dot Line', linewidth=2)
plt.plot(x, x+3, 'm:', label='Dotted Line', linewidth=2)
plt.plot(x, x+4, 'co-', label='Circles', markersize=8)
plt.plot(x, x+5, 's-', label='Squares', markersize=6)
plt.plot(x, x+6, '^-', label='Triangles', markersize=6)

plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Line Styles and Markers')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Using Colors, names and hex codes
x = np.linspace(0, 10, 100)
plt.figure(figsize=(10, 6))
plt.plot(x, np.sin(x), color='steelblue', label='Steelblue', linewidth=2)
plt.plot(x, np.cos(x), color='#FF6B6B', label='#FF6B6B (hex)', linewidth=2)
plt.plot(x, np.sin(x)*0.5, color='darkgreen', label='Dark green dashed', linestyle='--')
plt.xlabel('X Values', fontsize=12)
plt.ylabel('Y Values', fontsize=12)
plt.title('Custom Colors and Styles', fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.show()
```

## Scatter Plots
### Scatter plots are perfect for visualizing relationships between two variables and identifying patterns or correlations.

```python
# Generate Sample data
np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + np.random.randn(100) * 0.5 # Linear relationship with noise
# Basic Scatter plot
plt.figure(figsize=(8,6))
plt.scatter(x, y, alpha=0.6)
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.title('Scatter plot')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Scatter plot with different colrs and sizes
np.random.seed(42)
x = np.random.randn(100)
y = np.random.randn(100)
colors = np.random.randn(100) # Random colors
sizes = 1000 * np.random.randn(100) # Random sizes


plt.figure(figsize=(10, 6))
plt.scatter(x, y, c=colors, s=sizes, alpha=0.6, cmap='Set1')
plt.colorbar(label='Color intensity') # Color bar
plt.xlabel('X Values')
plt.ylabel('Y Values')
plt.title('Scatter plot with Color and Size variations')
plt.grid(True, alpha=0.3)
plt.show()

# Polpular colormaps: 'viridis', 'plasma', 'coolwarm', 'Rdy1bu', 'Set1'
```

## Bar Charts
### Bar charts are ideal for comparing categorical data or showing counts

```python
# Simple bar charts
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 32]
plt.figure(figsize=(8, 6))
plt.bar(categories, values, color='steelblue', edgecolor='black', linewidth=1.2)
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Simple Bar Chart')
plt.grid(True, alpha=0.3, axis='y') # Only horizental  grid line
plt.show()
```

```python
# Horizental Bar Chart
plt.figure(figsize=(8, 6))
plt.barh(categories, values, color='coral', edgecolor='black', linewidth=1.2)
plt.xlabel('Values')
plt.ylabel('Categories')
plt.title('Horizental Bar Chart')
plt.grid(True, alpha=0.3, axis='x')
plt.show()
```

## Histogram

```python
# Generate sample data
np.random.seed(42)
data = np.random.normal(100, 15, 1000)
plt.figure(figsize=(10, 6))
plt.hist(data, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram of Normal Distribution')
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

## Pie Charts

```python
# Single Pie Chart
categories = ['North', 'South', 'East', 'West']
sizes = [30, 25, 20, 25]
colors = ['#FF6B6B', '#4ECDC4', '#95E1D3', '#F3B1B1']
explode = [0.1, 0, 0, 0] # Explode the first slice
fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(sizes, explode=explode, labels=categories, colors=colors, autopct='%1.1f%%', shadow=True, startangle=90)
ax.set_title('Sales Distribution by Region', fontsize=14, fontweight='bold')
plt.show()
```

## Saving Figures

```python
# Create a sample plot
x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(x, y, 'b-', linewidth=2)

ax.set_xlabel('X values')
ax.set_ylabel('sin(x)')
ax.set_title('Sample Plot for Saving')

ax.grid(True, alpha=0.3)

# Save the figure
# Common formats: PNG (raster), PDF (vector), SVG (vector), EPS (vector)

fig.savefig('my_plot.png', dpi=300, bbox_inches='tight')

fig.savefig('my_plot.pdf', bbox_inches='tight')  # Vector format (scalable)

print("Figure saved as 'my_plot.png' and 'my_plot.pdf'")

print("\nCommon save parameters:")

print("- dpi: Resolution (dots per inch). Higher = better quality but larger file")

print("- bbox_inches='tight': Removes extra whitespace")

print("- format: Explicit format specification")

print("- facecolor: Background color")

plt.show()
```
